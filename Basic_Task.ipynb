Task 4: Weekly Step Tracker

```python
total_steps = 0

for i in range (8):

  steps = int(input("Enter your todays steps: "))


  total_steps = total_steps + steps
print(f"total_steps now: ", total_steps)
average_steps = total_steps / 7

print(f"your average steps are : ", average_steps)

if average_steps >= 5000:
    print("Active week")
else:
    print ("Needs more activity")

```
