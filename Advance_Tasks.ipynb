**Task 11: Personal Expense Tracker**

```python
Expense = []
while True:
  Expense_name = input("Enter expense name: ")
  Category = input("Enter expense category: ")
  Amount = int(input("Enter expense amount: "))

  Expense_Dict = {
      "Expense_name" : Expense_name,
      "Category" : Category,
      "Amount" : Amount
  }
  Expense.append(Expense_Dict)

  print("Expense added successfully")

  choice = input("Do you have more expenses? (yes/no): ")
  if choice.lower() != "yes":
    break

print("\nExpense Summary:")
for Expense_Dict in Expense:
  print(f"  Expense Name: {Expense_Dict['Expense_name']}")
  print(f"  Category: {Expense_Dict['Category']}")
  print(f"  Amount: {Expense_Dict['Amount']}")

# Calculate Total Spendings for all expenses
Total_spendings = sum([item['Amount'] for item in Expense])
print(f"\nTotal Spendings: {Total_spendings}")

# Calculate Category-wise Spendings for all expenses
category_spendings = {}
for item in Expense:
    category = item['Category']
    amount = item['Amount']
    category_spendings[category] = category_spendings.get(category, 0) + amount

print("\nCategory-wise Spendings:")
for category, total in category_spendings.items():
    print(f"  {category}: {total}")

# Highest Expense
highest_expense = max(Expense, key=lambda x: x['Amount'])
print("\nHighest Expense:")
print(f"  Expense Name: {highest_expense['Expense_name']}")
print(f"  Category: {highest_expense['Category']}")
print(f"  Amount: {highest_expense['Amount']}")

User_budget = int(input("Enter your budget: "))
remaining_budget = User_budget - Total_spendings


if Total_spendings > User_budget:
  print("Warning! You have exceeded your budget")
else:
  print(f"Remaining Budget: {remaining_budget}")

```

**Task 12: Store Inventory Manager**

```python
products = []

def add_products():
  product_barcode = int(input("Enter the product's code: "))

  for product in products:
    if product["product_barcode"] == product_barcode:
      print("Product already exists")
      return
  product_name = input("Enter product name: ")
  product_price = int(input("Enter product price: "))
  product_quantity = int(input("Enter product quantity: "))


  products.append({
      "product_barcode" : product_barcode,
      "product_name" : product_name,
      "Product_price" : product_price,
      "Product_quantity" : product_quantity
  })
  print("Product added successfully")

def sell_product():
  product_barcode = int(input("Enter product barcode: "))

  for product in products:
        if product["product_barcode"] == product_barcode:

            quantity_to_sell = int(input("Enter quantity to sell: "))

            if product["Product_quantity"] >= quantity_to_sell:
                product["Product_quantity"] -= quantity_to_sell
                print("Product sold successfully")
            else:
                print("Insufficient stock")

            return

  print("Product not found")

def restock_product():
  product_barcode = int(input("Enter product barcode: "))

  for product in products:
        if product["product_barcode"] == product_barcode:

            quantity_to_add = int(input("Enter quantity to add: "))

            product["Product_quantity"] += quantity_to_add

            print("Product restocked successfully")
            return

  print("Product not found")




def search_product():
    product_barcode = int(input("Enter product Barcode: "))

    for product in products:
        if product["product_barcode"] == product_barcode:
            print(f"Product name: {product['product_name']}")
            print(f"Product price: {product['Product_price']}")
            print(f"Product quantity: {product['Product_quantity']}")

            if product["Product_quantity"] < 5:
                print("Warning: Low stock!")

            return

    print("Product not found")

def display_inventory():
  if len(products) == 0:
        print("Inventory is empty")
        return

  total_inventory_value = 0

  print("----- Inventory -----")

  for product in products:
        product_value = (
            product["Product_price"] *
            product["Product_quantity"]
        )

        total_inventory_value += product_value

        print(f"Product code: {product['product_barcode']}")
        print(f"Product name: {product['product_name']}")
        print(f"Product price: {product['Product_price']}")
        print(f"Product quantity: {product['Product_quantity']}")
        print(f"Product value: {product_value}")

        if product["Product_quantity"] < 5:
            print("Status: LOW STOCK")

        print("--------------------")

  print(f"Total Inventory Value: {total_inventory_value}")


while True:

  print(" 1. Add Product")
  print(" 2. Sell Product")
  print(" 3. Restock Prodct")
  print(" 4. Search Product")
  print(" 5. Display Inventory")
  print(" 6. Exit")


  choice = int(input("Enter your choice: "))

  if choice == 1:
    add_products()

  elif choice == 2:
    sell_product()

  elif choice == 3:
    restock_product()

  elif choice == 4:
    search_product()

  elif choice == 5:
    display_inventory()

  elif choice == 6:
    break

  else:
    print("Invalid choice")
```

**Task 13: Library Book Management System**

```python
books = []


def add_book():
    book_id = int(input("Enter book ID: "))

    # Check if book ID already exists
    for book in books:
        if book["book_id"] == book_id:
            print("Book ID already exists!")
            return

    book_title = input("Enter book title: ")
    author = input("Enter author name: ")

    books.append({
        "book_id": book_id,
        "book_title": book_title,
        "author": author,
        "available": True
    })

    print("Book added successfully!")


def display_books():
    if len(books) == 0:
        print("No books available in library.")
        return

    print("----- Library Books -----")

    for book in books:
        print(f"Book ID: {book['book_id']}")
        print(f"Book Title: {book['book_title']}")
        print(f"Author: {book['author']}")

        if book["available"]:
            print("Status: Available")
        else:
            print("Status: Borrowed")

        print("------------------------")


def search_book():
    title = input("Enter book title to search: ")

    for book in books:
        if book["book_title"].lower() == title.lower():
            print("\nBook Found!")
            print(f"Book ID: {book['book_id']}")
            print(f"Book Title: {book['book_title']}")
            print(f"Author: {book['author']}")

            if book["available"]:
                print("Status: Available")
            else:
                print("Status: Borrowed")

            return

    print("Book not found!")


def borrow_book():
    book_id = int(input("Enter book ID to borrow: "))

    for book in books:
        if book["book_id"] == book_id:

            if book["available"]:
                book["available"] = False
                print("Book borrowed successfully!")
            else:
                print("Book is already borrowed.")

            return

    print("Book not found!")


def return_book():
    book_id = int(input("Enter book ID to return: "))

    for book in books:
        if book["book_id"] == book_id:

            if book["available"]:
                print("This book is already available.")
            else:
                book["available"] = True
                print("Book returned successfully!")

            return

    print("Book not found!")


while True:

    print("===== Library Book Management System =====")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        add_book()

    elif choice == 2:
        display_books()

    elif choice == 3:
        search_book()

    elif choice == 4:
        borrow_book()

    elif choice == 5:
        return_book()

    elif choice == 6:
        print("Program ended. Goodbye!")
        break

    else:
        print("Invalid choice!")
```

**Task 14: Restaurant Ordering system**

```python
menu = {
    "Burger" : 500,
    "Pizza" : 1200,
    "Pasta" : 800,
    "Drink" : 200
}
orders = []

print("======= Restaurant Menu =======")
for items, price in menu.items():
  print(f"{items} : {price}")

while True:

  item = input("Enter the item you want to order: ").capitalize()


  if item in menu:
    quantity = int(input(f"Enter the quantity of {item}: "))
    item_subtotal = menu[item] * quantity

    orders.append({"items" : item,
                   "quantity" : quantity,
                   "subtotal" : item_subtotal
                   })


  else:
    print("Item not available")

  choice = input("Do you want to order more items? (yes/no): ")
  if choice == "no":
   break

total = 0

for order in orders:
  total = total + order["subtotal"]

tax = total * 0.13

if total >= 3000:
  discount = total * 0.1

else:
  discount = 0

grand_total = total + tax - discount


def receipt():
  print("=========Receipt=========")

for order in orders:
  print(f'{order["items"]} x {order["quantity"]} = {order["subtotal"]}')

print("Total", total)
print("Tax", tax)
print("Discount", discount)
print("Grand Total", grand_total)
```

**Task 15: Clinic Appointment Manager**

```python
patients = []
def add_patient():
  patient_id = int(input("Enter  you medical id: "))

  for patient in patients:
    if patient["patient_id"] == patient_id:
      print("Patient already exists")
      return

  patient_name = input("Enter patient name: ")
  patient_age = int(input("Enter patient age: "))
  problem = input("Enter patient problem: ")
  time = float(input("Enter appointment time: "))
  emergency = input("Is there any emergency with patient? (yes/No): ")

  patients.append({
      "patient_id" : patient_id,
      "patient_name" : patient_name,
      "patient_age" : patient_age,
      "problem" : problem,
      "time" : time,
      "emergency" : emergency
  })
  print("Patient added successfully")

def search_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["patient_id"] == patient_id:
      print(f"Patient name: {patient['patient_name']}")
      print(f"Patient age: {patient['patient_age']}")
      print(f"Patient problem: {patient['problem']}")
      print(f"Appointment time: {patient['time']}")

      return

    else:
      print("Patient not found")

def display_appointment():
  for patient in patients:
    if patient["emergency"] == "yes":
      print(f"Patient name: {patient['patient_name']}")
      print(f"Patient age: {patient['patient_age']}")
      print(f"Patient problem: {patient['problem']}")
      print(f"Appointment time: {patient['time']}")



def count_patient():
  count = len(patients)
  print(f"Total number of patients: {count}")

def delete_patient():
  patient_id = int(input("Enter patient id: "))
  for patient in patients:
    if patient["patient_id"] == patient_id:
      patients.remove(patient)
      print("Patient deleted successfully")
      return



while True:

  print(" 1. Add Patient")
  print(" 2. Search Patients")
  print(" 3. Display Appointment")
  print(" 4. Count Patient")
  print(" 5. Delete Patient")
  print(" 6. Exit")


  choice = int(input("Enter your choice: "))

  if choice == 1:
    add_patient()

  elif choice == 2:
    search_patient()

  elif choice == 3:
    display_appointment()

  elif choice == 4:
    count_patient()

  elif choice == 5:
    delete_patient()

  elif choice == 6:
    break

  else:
    print("Invalid choice")
```
